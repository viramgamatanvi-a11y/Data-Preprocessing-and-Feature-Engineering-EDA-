## 1️⃣ What is Data Analysis?

### 📌 Answer:

Data Analysis is the process of collecting, cleaning, organizing, and examining data to find useful information, patterns, and insights. It helps businesses and organizations make better decisions based on data.

### 🎯 Purpose of Data Analysis:
- Understand data clearly
- Find trends and patterns
- Solve business problems
- Support decision-making
- Improve performance

### 💡 Example:

A shopping company analyzes customer purchase data to find which products are sold the most. Based on this analysis, the company can increase the stock of popular products.

## 2️⃣ Steps to Plan a Data Science Project

### 📌 Answer:

A Data Science project follows a structured process from understanding the problem to deploying the final model.

### 🔄 Steps of a Data Science Project:

### 1. Business Understanding
- Understand the problem.
- Define project goals.

### 2. Data Collection
- Gather data from databases, files, APIs, or websites.

### 3. Data Understanding
- Explore the dataset.
- Identify data types and missing values.

### 4. Data Cleaning
- Handle missing values.
- Remove duplicates.
- Fix incorrect data.

### 5. Exploratory Data Analysis (EDA)
- Analyze data using statistics and visualizations.
- Find patterns and relationships.

### 6. Feature Engineering
- Create new useful features.
- Transform existing features.

### 7. Model Building
- Select and train Machine Learning models.

### 8. Model Evaluation
- Measure model performance using evaluation metrics.

### 9. Model Deployment
- Deploy the model for real-world use.

### 10. Monitoring and Maintenance
- Monitor model performance.
- Update the model when required.

### 📊 Example:

House Price Prediction

- Problem: Predict house prices.
- Data Collection: Gather housing data.
- EDA: Analyze house features.
- Model Training: Train a regression model.
- Deployment: Create a web application for predictions.

## 3️⃣ Machine Learning Problem Statement

### 📌 Problem Statement:

Predict whether a customer will churn based on their purchase behavior and transaction history.

### 🎯 Objective:

The goal is to identify customers who are likely to stop using the company's products or services.

### 📥 Input Features:

- Purchase Frequency
- Total Spending
- Last Purchase Date
- Customer Age
- Membership Duration
- Number of Transactions

### 📤 Output:

- Churn = Yes
- Churn = No

### 🤖 Type of Machine Learning Problem:

**Classification Problem**

Because the output belongs to one of two categories:
- Churn
- Not Churn

### 💡 Example:

| Purchase Frequency | Total Spending | Churn |
|-------------------|----------------|--------|
| High | ₹50,000 | No |
| Low | ₹2,000 | Yes |

The model learns from past customer behavior and predicts future churn.

## 4️⃣ What are Tensors?

### 📌 Answer:

A Tensor is a mathematical structure used to store data in multiple dimensions.

A tensor is a generalization of:
- Scalar (0D)
- Vector (1D)
- Matrix (2D)
- Higher-Dimensional Arrays (3D and above)

Tensors are widely used in:
- Machine Learning
- Deep Learning
- Artificial Intelligence
- Computer Vision
- Natural Language Processing
